```python
#Removing inconsistency
#one of the way by which inconsistencies are removed is using the strip library, which basically act like a trimming function.
#However, it is important we know that there are 2 common style of striping, these are:
#lstrip = left stripping 
#Rstrip = right stripping
```

```python
df["Last_Name"].str.strip()
```

```python
df["Last_Name"].str.lstrip("...")
```

```python
df["Last_Name"]= df["Last_Name"].str.lstrip("...")
```

```python
df
```

```python
df["Last_Name"]= df["Last_Name"].str.lstrip("/")
df["Last_Name"]= df["Last_Name"].str.lstrip("_")
```

```python
df
```

```python
df["Last_Name"]= df["Last_Name"].str.rstrip("_")
```

```python
df
```

```python
df["Last_Name"]= df["Last_Name"].str.strip("123./ _")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace(r"[^0-9]","")
```

```python
df
```

```python
df["Phone_Number"].str.replace(r"[^a-zA-z 0-9]","")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace(r"[^a-zA-Z0-9]","", regex= True)
```

```python
df
```

```python
#How to convert a column to string. i.e from int to str
# note, you need to specify the column you are changing then use lamda to correct that or simply  use astype("string") 
# in short term, lamda is the function to get the task done
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x:str(x))
```

```python
df["Phone_Number"] = df["Phone_Number"].astype("string")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x:x[0:3] + "-" + x[3:6] + "-" + x[6:10])
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace("nan--","")
```

```python
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace("Na--","")
```
